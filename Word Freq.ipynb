```python
from wordfreq import word_frequency as freq
from wordfreq import zipf_frequency as zfreq
import numpy as np
import csv as csv
```

```python
filename='stimulus homonyms.csv'
with open(filename,mode='r',newline='',encoding='utf-8') as file:
    reader=csv.reader(file)
    for row in reader:
        if abs(zfreq(row[2],'en')-zfreq(row[4],'en'))>0.3:
        
            print(f"{row[2]},{row[4]}:{zfreq(row[2],'en')-zfreq(row[4],'en')}, chat's difference={row[6]}")
```

```python
import numpy as np
import pandas as pd
import csv
import glob as glob
from statistics import NormalDist as norm
import matplotlib.pyplot as plt
```

```python
def read_to_df(filename):
    ret=pd.DataFrame(columns=['H','SR0', 'SR1', 'UNH', 'PH','PNH'],dtype='int64')
    responses=pd.read_csv(filename)['resp']
    #print(responses)
    for i in range(0,len(responses),6):
        #print(i//6)
        #print(responses[i:i+6])
        ret.loc[i//6]=[resp for resp in responses[i:i+6]]
    return ret

def get_rates(arr):
    ret=np.array([0,0,0])
    total=0
    for resp in arr:
        total+=1
        if resp>1:
            ret[:resp]+=1
    return ret/total
```

```python
#print(glob.glob('*'))
files='data_csvs/Condition 0*-responses.csv'
ret=[0,0,0]
total=0
rates=[{} for i in range(len(glob.glob(files)))]
zs=[{} for i in range(len(glob.glob(files)))]

d_prime_difference=[0,0,0]
control_d_prime=[0,0,0]

for i,filename in enumerate(glob.glob(files)):
    df=read_to_df(filename)
    
    for col in df.columns:
        r=get_rates(df[col])
        z=np.array([norm().inv_cdf(x) for x in r])
        rates[i][col]=r
        zs[i][col]=z
    d_prime_difference+=zs[i]['SR1']-zs[i]['SR0']
    control_d_prime+=zs[i]['PNH']-zs[i]['PH']
    

print([x/(i+1) for x in d_prime_difference])
print([x/(i+1) for x in control_d_prime])
```

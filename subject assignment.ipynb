```python
import random as r
import csv
import pandas as pd
```

```python
r.seed('hello world')
subs = ["vidita", "leilani", "wesley", "anna", "sophia", "erica", "jiarong","jackson", "rebecca", "patrick", "ava", "hannah", "john", "vy", "david", "allah", "riley"]


# Shuffle subs
r.shuffle(subs)

# Put first 10 subs in 3-target condition, next 3 subs in the 1-target-animal, next 3 in 1-target-tool, next 3 in 1-target-produce
meaning_1 = subs[0:8]
meaning_2= subs[8:17]

# Let's find out where each sub goes:
print('meaning_1 condition:')
print(meaning_1)
print('meaning_2 condition')
print(meaning_2)
```

```python
data_array=[]
for sub in subs:
    if sub in meaning_1:
        data_array.append([sub,0])
    elif sub in meaning_2:
        data_array.append([sub,1])
    else:
        raise ValueError(f'Subject \"{sub}\" not found')

data_df=pd.DataFrame(data_array,columns=["Participant Name","Condition"])
output_filename='conditions.csv'
data_df.to_csv(output_filename)
```

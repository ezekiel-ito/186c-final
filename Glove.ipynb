```python
import gensim.downloader as api
import wordfreq
from wordfreq import word_frequency as freq
from wordfreq import zipf_frequency as zfreq
import csv
import numpy as np

# Downloads and loads the 100-dimensional Wikipedia + Gigaword GloVe model
glove_model = api.load("glove-wiki-gigaword-100")

```

```python
# Retrieve the vector for a specific word
vector = glove_model["king"]

# Find semantically similar words
similar_words = glove_model.most_similar("king")
print(similar_words)
```

```python
v1=glove_model["bank"]
v2=glove_model["river"]
v3=glove_model["atm"]
#print(v1,v2,v3)
print(np.linalg.norm(v1-v2))
print(np.linalg.norm(v1-v3))
```
